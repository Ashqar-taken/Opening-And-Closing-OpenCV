```python
# Import the necessary packages
import cv2
import numpy as np
from matplotlib import pyplot as plt
     
```

```python
# Create the text using cv2.putText
img = np.zeros((500, 500 ,3), dtype=np.uint8)
cv2.putText(img, 'OPENING and CLOSING', (100, 250), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 3)
```

```python
# Display the image with text:
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for displaying
plt.title("Input Image with Text")
plt.axis('off')
```

```python
# Create the structuring element
kernel = np.ones((3,3), np.uint8)
```

```python
# Use the Opening Operation
img_opening = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
```

```python
plt.imshow(cv2.cvtColor(img_opening, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for displaying
plt.title("Image after Opening")   
plt.axis('off')
```

```python
# Use the Closing Operation
img_closing = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)
```

```python
plt.imshow(cv2.cvtColor(img_closing, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for displaying
plt.title("Image after Closing")
plt.axis('off')
```
